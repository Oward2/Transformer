## 3. Loading and preprocessing the data

ViTs divide images into fixed-size patches, so image dimensions must be divisible by the patch size. This template keeps grayscale inputs to match the original notebook. For RGB, remove the grayscale transform and set `IN_CHANNELS = 3`.

````python
# Data processing Values, such as image and patch size
IMG_SIZE = 120
PATCH_SIZE = 6
IN_CHANNELS = 1
BATCH_SIZE = 64
VAL_FRACTION = 0.20
NUM_WORKERS = 5

assert IMG_SIZE % PATCH_SIZE == 0

#############################################################################
# Data augmentation and Transforms

baseline_transform = transforms.Compose([
    transforms.Grayscale(num_output_channels=IN_CHANNELS),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5], std=[0.5]),
])

# Adjust augmentation as part of your experiments.
train_transform = transforms.Compose([
    transforms.Grayscale(num_output_channels=IN_CHANNELS),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomResizedCrop(IMG_SIZE, scale=(0.8, 1.0), ratio=(0.75, 1.333)),
    transforms.RandomHorizontalFlip(p=0.5),
     transforms.ColorJitter(brightness=0.1,contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5], std=[0.5]),
])


hard_transform = transforms.Compose([
    transforms.Grayscale(num_output_channels=1),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ColorJitter(
        brightness=0.3,
        contrast=0.6
    ),
    transforms.RandomResizedCrop(
        size=(IMG_SIZE, IMG_SIZE),
        scale=(.8, 1.0)
    ),
    transforms.RandomVerticalFlip(.5),
    transforms.RandomHorizontalFlip(p=.5),
    transforms.ToTensor(),
    transforms.RandomErasing(
        p=0.7,
        scale=(0.1, 0.30),
        ratio=(1, 1)
    ),
     transforms.RandomSolarize(.9, .4),
     # transforms.ColorJitter(
    #     brightness=0.5,
    #     contrast=0.5,
    # ),

    transforms.Normalize(mean=[0.5], std=[0.5]),

])
###############################################################################################
# Initialize data set configurations

full_train_dataset = datasets.ImageFolder(TRAIN_ROOT, transform=baseline_transform)
test_dataset = datasets.ImageFolder(TEST_ROOT, transform=baseline_transform)
class_names = full_train_dataset.classes
num_classes = len(class_names)


###############################################################################################
# Same split procedure as the original notebook:
# create seeded indices, then apply them to separate transformed datasets.
val_size = int(round(len(full_train_dataset) * VAL_FRACTION))
train_size = len(full_train_dataset) - val_size
generator = torch.Generator().manual_seed(SEED)

train_subset, val_subset = random_split(
    full_train_dataset, [train_size, val_size], generator=generator
)

train_full = datasets.ImageFolder(TRAIN_ROOT, transform=train_transform)
val_full = datasets.ImageFolder(TRAIN_ROOT, transform=baseline_transform)

train_dataset = torch.utils.data.Subset(train_full, train_subset.indices)
val_dataset = torch.utils.data.Subset(val_full, val_subset.indices)

assert test_dataset.class_to_idx == full_train_dataset.class_to_idx, (
    "Train and test class mappings do not match."
)

################################################################################################
train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)

################################################################################
# SimCLR pretraining set (labels are never used for pretraining)

SIMCLR_BATCH_SIZE = 128      # contrastive learning benefits from larger batches; lower if you run out of memory

# Grayscale-friendly SimCLR augmentations. Hue/saturation jitter and random grayscale
# are omitted because the inputs are already single-channel.
simclr_transform = transforms.Compose([
    transforms.Grayscale(num_output_channels=IN_CHANNELS),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomResizedCrop(IMG_SIZE, scale=(0.3, 1.0), ratio=(0.75, 1.333)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomApply([transforms.ColorJitter(brightness=0.4, contrast=0.4)], p=0.8),
    transforms.RandomApply([transforms.GaussianBlur(kernel_size=9, sigma=(0.1, 2.0))], p=0.5),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5], std=[0.5]),
])


# DataLoader workers started with "spawn"/"forkserver" (the default on Python 3.14 / Linux and on
# Windows/macOS) cannot unpickle classes defined inside a notebook's __main__. Keeping the dataset
# class in a real module file makes multi-worker loading work on every platform.
import sys
Path("simclr_data.py").write_text('''
import torch

class TwoViewDataset(torch.utils.data.Dataset):
    """Return two independently augmented views of the same image (label is ignored)."""
    def __init__(self, base_dataset, transform):
        self.base = base_dataset          # ImageFolder with transform=None (returns PIL images)
        self.transform = transform

    def __len__(self):
        return len(self.base)

    def __getitem__(self, idx):
        image, label = self.base[idx]
        return self.transform(image), self.transform(image), label
''')
sys.path.insert(0, str(Path(".").resolve()))
from simclr_data import TwoViewDataset


# Every training-folder image, two augmented views each
simclr_raw = datasets.ImageFolder(TRAIN_ROOT, transform=None)
simclr_train_dataset = TwoViewDataset(simclr_raw, simclr_transform)

# Monitoring set: the validation indices, also two-view. This only tracks the
# contrastive loss; these images were also used (unlabeled) during pretraining.
simclr_monitor_dataset = TwoViewDataset(
    torch.utils.data.Subset(simclr_raw, val_subset.indices), simclr_transform
)

# Plain (non-augmented) datasets for the k-NN quality probe during pretraining
knn_train_dataset = torch.utils.data.Subset(val_full, train_subset.indices)
knn_val_dataset = val_dataset

simclr_train_loader = DataLoader(
    simclr_train_dataset, batch_size=SIMCLR_BATCH_SIZE, shuffle=True, drop_last=True,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)
simclr_monitor_loader = DataLoader(
    simclr_monitor_dataset, batch_size=SIMCLR_BATCH_SIZE, shuffle=False, drop_last=True,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)
knn_train_loader = DataLoader(
    knn_train_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)
knn_val_loader = DataLoader(
    knn_val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)

print(f"SimCLR pretraining images: {len(simclr_train_dataset)} "
      f"(was {len(train_dataset)} train + {len(val_dataset)} val)")

print("Classes:", class_names)
print(f"Train: {len(train_dataset)} | Validation: {len(val_dataset)} | Test: {len(test_dataset)}")
````

### Dataset summary tables

The functions below display per-class counts and total image counts for each split.

````python
def get_targets(dataset):
    """Get labels from ImageFolder or Subset(ImageFolder)."""
    if isinstance(dataset, torch.utils.data.Subset):
        targets = np.asarray(dataset.dataset.targets)
        return targets[np.asarray(dataset.indices)]
    if hasattr(dataset, "targets"):
        return np.asarray(dataset.targets)
    raise TypeError("Expected ImageFolder or Subset(ImageFolder).")


def display_dataset_tables(train_dataset, val_dataset, test_dataset, class_names):
    """Display per-class and per-split image counts as pandas tables."""
    split_datasets = {
        "Train": train_dataset,
        "Validation": val_dataset,
        "Test": test_dataset
    }
    rows = []
    for class_id, class_name in enumerate(class_names):
        row = {"Class": class_name}
        for split_name, dataset in split_datasets.items():
            row[split_name] = int(np.sum(get_targets(dataset) == class_id))
        rows.append(row)

    class_table = pd.DataFrame(rows)
    totals = {"Class": "TOTAL"}
    for split_name, dataset in split_datasets.items():
        totals[split_name] = len(dataset)
    class_table.loc[len(class_table)] = totals

    split_table = pd.DataFrame([
        {"Split": name, "Images": len(dataset)}
        for name, dataset in split_datasets.items()
    ])

    print("Images per class and split")
    display(class_table)
    print("Dataset split summary")
    display(split_table)
    return class_table, split_table


class_count_table, split_summary_table = display_dataset_tables(
    train_dataset, val_dataset, test_dataset, class_names
)
````

### Visualize training examples

The helper reverses normalization before displaying images.

````python
def show_image_batch(loader, class_names, max_images=8, mean=0.5, std=0.5):
    images, labels = next(iter(loader))
    count = min(max_images, len(images))
    cols = 4
    rows = math.ceil(count / cols)
    fig, axes = plt.subplots(rows, cols, figsize=(12, 3 * rows))
    axes = np.asarray(axes).reshape(-1)

    for ax in axes:
        ax.axis("off")

    for ax, image, label in zip(axes, images[:count], labels[:count]):
        image = (image * std + mean).clamp(0, 1)
        if image.shape[0] == 1:
            ax.imshow(image.squeeze(0), cmap="gray")
        else:
            ax.imshow(image.permute(1, 2, 0))
        ax.set_title(class_names[label.item()])
        ax.axis("off")

    plt.tight_layout()
    plt.show()

show_image_batch(train_loader, class_names)
````

### Visualize SimCLR augmented views

Each row shows one image and two random augmentations of it. SimCLR trains the encoder to pull these two views together in embedding space and push them away from the views of every other image in the batch.

````python
def show_simclr_views(dataset, n=4, mean=0.5, std=0.5):
    fig, axes = plt.subplots(n, 3, figsize=(8, 2.7 * n), squeeze=False)
    for r in range(n):
        idx = random.randrange(len(dataset))
        original, _ = dataset.base[idx]
        view1, view2, _ = dataset[idx]
        panels = [("Original", original), ("View 1", view1), ("View 2", view2)]
        for c, (title, img) in enumerate(panels):
            ax = axes[r, c]
            if torch.is_tensor(img):
                img = (img * std + mean).clamp(0, 1)
                ax.imshow(img.squeeze(0) if img.shape[0] == 1 else img.permute(1, 2, 0),
                          cmap="gray" if img.shape[0] == 1 else None, vmin=0, vmax=1)
            else:
                ax.imshow(img, cmap="gray" if img.mode == "L" else None)
            ax.axis("off")
            if r == 0:
                ax.set_title(title)
    plt.tight_layout()
    plt.show()

show_simclr_views(simclr_train_dataset)
````

## 4. Vision Transformer model

The model converts image patches into embeddings, adds a learnable class token and positional embeddings, processes the token sequence with Transformer encoder layers, and classifies using the class-token output.

For SimCLR, the ViT acts as the **encoder** `f(·)`. Its class-token output `h` is passed through a small **projection head** `g(·)` to produce `z`, and the contrastive loss is applied to `z`. After pretraining, the projection head is discarded and a linear classification head is attached to the encoder.

````python
# This class acts as the Tokenizer creating vector representations of the patches generated per Convolution
class PatchEmbedding(nn.Module):
    """Convert an image into a sequence of patch embeddings."""
    def __init__(self, img_size=128, patch_size=16, in_channels=1, embed_dim=192):
        super().__init__()
        assert img_size % patch_size == 0
        self.num_patches = (img_size // patch_size) ** 2
        self.projection = nn.Conv2d(
            in_channels, embed_dim,
            kernel_size=patch_size, stride=patch_size
        )

    def forward(self, x):
        x = self.projection(x)       # [B, embed_dim, patch_rows, patch_cols]
        x = x.flatten(2)             # [B, embed_dim, patch_count]
        return x.transpose(1, 2)     # [B, patch_count, embed_dim]


class VisionTransformer(nn.Module):

    def __init__(
        self, img_size=128, patch_size=16, in_channels=1,
        num_classes=16, embed_dim=128, num_heads=4,
        num_layers=6, mlp_dim=384, dropout=0.1
    ):
        super().__init__()
        self.embed_dim = embed_dim

        # Stage 1: Patch embedding
        self.patch_embed = PatchEmbedding(
            img_size, patch_size, in_channels, embed_dim
        )
        num_patches = self.patch_embed.num_patches

        # Stage 2: Token preparation
        self.class_token = nn.Parameter(torch.zeros(1, 1, embed_dim))
        self.position_embedding = nn.Parameter(torch.zeros(1, num_patches + 1, embed_dim))
        self.embedding_dropout = nn.Dropout(dropout)

        # Stage 3: Transformer encoder
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim,
            nhead=num_heads,
            dim_feedforward=mlp_dim,
            dropout=dropout,
            activation="gelu",
            batch_first=True,
            norm_first=False
        )
        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers,
            norm=nn.LayerNorm(embed_dim)
        )

        # Stage 4: Classification (unused during SimCLR pretraining)
        self.classifier = nn.Linear(embed_dim, num_classes)

        nn.init.trunc_normal_(self.class_token, std=0.02)
        nn.init.trunc_normal_(self.position_embedding, std=0.02)

    def prepare_tokens(self, x):
        x = self.patch_embed(x)
        class_token = self.class_token.expand(x.shape[0], -1, -1)
        x = torch.cat((class_token, x), dim=1)
        x = x + self.position_embedding
        return self.embedding_dropout(x)

    def forward_features(self, x):
        """Return the class-token representation h: [B, embed_dim]."""
        x = self.prepare_tokens(x)
        x = self.transformer(x)
        return x[:, 0]

    def forward(self, x):
        return self.classifier(self.forward_features(x))


class ProjectionHead(nn.Module):
    """SimCLR projection head g(.): maps encoder features h to the contrastive space z."""
    def __init__(self, in_dim, hidden_dim=512, out_dim=128):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(inplace=True),
            nn.Linear(hidden_dim, out_dim),
        )

    def forward(self, h):
        return self.net(h)


class SimCLR(nn.Module):
    """ViT encoder + projection head. forward() returns the (unnormalized) projection z."""
    def __init__(self, vit_model, proj_hidden_dim=512, proj_out_dim=128):
        super().__init__()
        self.encoder = vit_model
        self.projection_head = ProjectionHead(
            vit_model.embed_dim, proj_hidden_dim, proj_out_dim
        )

    def forward(self, images):
        h = self.encoder.forward_features(images)
        return self.projection_head(h)


class SimCLRClassifier(nn.Module):
    """Pretrained SimCLR encoder + linear classification head (projection head is dropped)."""
    def __init__(self, simclr_model, num_classes):
        super().__init__()
        self.encoder = simclr_model.encoder          # shared with the SimCLR model
        self.classifier = nn.Linear(self.encoder.embed_dim, num_classes)

    def forward(self, images):
        return self.classifier(self.encoder.forward_features(images))


EMBED_DIM = 128
NUM_LAYERS = 6
NUM_HEADS = 8
MLP_DIM = 512

DROPOUT = 0
model = VisionTransformer(
    img_size=IMG_SIZE, patch_size=PATCH_SIZE,
    in_channels=IN_CHANNELS, num_classes=num_classes,
    embed_dim=EMBED_DIM, num_heads=NUM_HEADS,
    num_layers=NUM_LAYERS, mlp_dim=MLP_DIM, dropout=DROPOUT
).to(device)

print(model)
print("Trainable parameters:", sum(p.numel() for p in model.parameters() if p.requires_grad))
````

## 5. Training and evaluation utilities

The SimCLR utilities include the NT-Xent contrastive loss, a k-NN probe for monitoring representation quality, and a pretraining loop. The classifier utilities track loss and accuracy on both training and validation data and restore the checkpoint with the best validation accuracy.

````python
def nt_xent_loss(z1, z2, temperature=0.5):
    """
    NT-Xent (normalized temperature-scaled cross entropy) loss from SimCLR.

    z1, z2: [B, D] projections of two views of the same B images.
    For each of the 2B samples, the positive is the other view of the same image
    and the other 2B - 2 samples in the batch are negatives.
    Returns (loss, contrastive_top1_accuracy).
    """
    B = z1.shape[0]
    z = F.normalize(torch.cat([z1, z2], dim=0), dim=1)       # [2B, D]

    logits = (z @ z.T) / temperature                          # cosine similarity / T
    logits = logits.masked_fill(
        torch.eye(2 * B, dtype=torch.bool, device=z.device), float("-inf")
    )                                                          # a sample is not its own negative

    # Index of the positive partner for each sample: i <-> i + B
    targets = torch.cat([torch.arange(B, 2 * B), torch.arange(0, B)]).to(z.device)

    loss = F.cross_entropy(logits, targets)
    accuracy = (logits.argmax(dim=1) == targets).float().mean()
    return loss, accuracy


@torch.no_grad()
def extract_features(encoder, loader, device):
    encoder.eval()
    feats, labels = [], []
    for images, y in loader:
        feats.append(F.normalize(encoder.forward_features(images.to(device)), dim=1).cpu())
        labels.append(y)
    return torch.cat(feats), torch.cat(labels)


@torch.no_grad()
def knn_accuracy(encoder, train_loader, val_loader, device, k=20, temperature=0.1):
    """Weighted k-NN accuracy on frozen encoder features (labels used only for monitoring)."""
    train_feats, train_labels = extract_features(encoder, train_loader, device)
    val_feats, val_labels = extract_features(encoder, val_loader, device)

    sims = val_feats @ train_feats.T                           # [Nv, Nt]
    top_sims, top_idx = sims.topk(min(k, train_feats.shape[0]), dim=1)
    top_labels = train_labels[top_idx]
    weights = (top_sims / temperature).exp()

    scores = torch.zeros(val_feats.shape[0], int(train_labels.max()) + 1)
    scores.scatter_add_(1, top_labels, weights)
    return (scores.argmax(dim=1) == val_labels).float().mean().item() * 100


def train_simclr(
    model,
    train_loader,
    val_loader,
    optimizer,
    scheduler=None,
    epochs=100,
    temperature=0.5,
    knn_every=10,
    device="cuda"
):
    history = {
        "train_loss": [], "val_loss": [],
        "train_contrastive_acc": [], "val_contrastive_acc": [],
        "knn_epoch": [], "knn_acc": [],
        "learning_rate": []
    }
    start = time.time()

    for epoch in range(epochs):

        # -------------------------
        # Training
        # -------------------------
        model.train()
        running_loss, running_acc, seen = 0.0, 0.0, 0

        for view1, view2, _ in train_loader:
            view1 = view1.to(device, non_blocking=True)
            view2 = view2.to(device, non_blocking=True)

            optimizer.zero_grad(set_to_none=True)

            # One forward pass over both views so BatchNorm sees all 2B samples
            z = model(torch.cat([view1, view2], dim=0))
            z1, z2 = z.chunk(2, dim=0)

            loss, acc = nt_xent_loss(z1, z2, temperature)
            loss.backward()
            optimizer.step()

            n = view1.size(0)
            running_loss += loss.item() * n
            running_acc += acc.item() * n
            seen += n

        train_loss = running_loss / seen
        train_acc = 100 * running_acc / seen

        # -------------------------
        # Validation (contrastive loss on augmented val images)
        # -------------------------
        model.eval()
        running_loss, running_acc, seen = 0.0, 0.0, 0

        with torch.no_grad():
            for view1, view2, _ in val_loader:
                view1 = view1.to(device, non_blocking=True)
                view2 = view2.to(device, non_blocking=True)

                z = model(torch.cat([view1, view2], dim=0))
                z1, z2 = z.chunk(2, dim=0)
                loss, acc = nt_xent_loss(z1, z2, temperature)

                n = view1.size(0)
                running_loss += loss.item() * n
                running_acc += acc.item() * n
                seen += n

        val_loss = running_loss / seen
        val_acc = 100 * running_acc / seen

        # -------------------------
        # Learning rate
        # -------------------------
        current_lr = optimizer.param_groups[0]["lr"]
        if scheduler is not None:
            scheduler.step()

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_contrastive_acc"].append(train_acc)
        history["val_contrastive_acc"].append(val_acc)
        history["learning_rate"].append(current_lr)

        # -------------------------
        # Optional k-NN probe
        # -------------------------
        knn_text = ""
        if knn_every and ((epoch + 1) % knn_every == 0 or epoch == epochs - 1):
            knn = knn_accuracy(model.encoder, knn_train_loader, knn_val_loader, device)
            history["knn_epoch"].append(epoch + 1)
            history["knn_acc"].append(knn)
            knn_text = f" | kNN Acc: {knn:.2f}%"

        print(
            f"Epoch [{epoch + 1:02d}/{epochs}] | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Train Con-Acc: {train_acc:.2f}% | "
            f"Val Con-Acc: {val_acc:.2f}% | "
            f"LR: {current_lr:.6f}"
            f"{knn_text} | {time.time() - start:.0f}s"
        )

    return model, history


def plot_simclr_history(history):
    epochs = range(1, len(history["train_loss"]) + 1)
    fig, axes = plt.subplots(1, 3, figsize=(16, 4))

    axes[0].plot(epochs, history["train_loss"], label="Train")
    axes[0].plot(epochs, history["val_loss"], label="Validation")
    axes[0].set_title("NT-Xent loss")

    axes[1].plot(epochs, history["train_contrastive_acc"], label="Train")
    axes[1].plot(epochs, history["val_contrastive_acc"], label="Validation")
    axes[1].set_title("Contrastive top-1 accuracy (%)")

    if history["knn_acc"]:
        axes[2].plot(history["knn_epoch"], history["knn_acc"], marker="o")
    axes[2].set_title("k-NN accuracy on validation (%)")

    for ax in axes:
        ax.set_xlabel("Epoch")
        ax.grid(alpha=0.25)
    axes[0].legend()
    axes[1].legend()
    plt.tight_layout()
    plt.show()
````

````python
def train_classifier(
    model,
    train_loader,
    val_loader,
    optimizer,
    scheduler=None,
    epochs=30,
    device="cuda"
):

    criterion = nn.CrossEntropyLoss()

    history = {
        "train_loss": [],
        "val_loss": [],
        "train_acc": [],
        "val_acc": [],
        "learning_rate": []
    }

    for epoch in range(epochs):

        # Training
        model.train()

        train_loss = 0.0
        train_correct = 0
        train_total = 0

        for images, labels in train_loader:

            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            optimizer.zero_grad(set_to_none=True)

            outputs = model(images)
            loss = criterion(outputs, labels)

            loss.backward()
            optimizer.step()

            train_loss += loss.item() * images.size(0)

            predictions = outputs.argmax(dim=1)

            train_correct += (predictions == labels).sum().item()
            train_total += labels.size(0)

        train_loss /= train_total
        train_accuracy = 100 * train_correct / train_total

        # Validation
        model.eval()

        val_loss = 0.0
        val_correct = 0
        val_total = 0

        with torch.no_grad():

            for images, labels in val_loader:

                images = images.to(device, non_blocking=True)
                labels = labels.to(device, non_blocking=True)

                outputs = model(images)
                loss = criterion(outputs, labels)

                val_loss += loss.item() * images.size(0)

                predictions = outputs.argmax(dim=1)

                val_correct += (predictions == labels).sum().item()
                val_total += labels.size(0)

        val_loss /= val_total
        val_accuracy = 100 * val_correct / val_total

        current_lr = optimizer.param_groups[0]["lr"]

        if scheduler is not None:
            scheduler.step()

        # Save history
        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(train_accuracy)
        history["val_acc"].append(val_accuracy)
        history["learning_rate"].append(current_lr)

        print(
            f"Epoch [{epoch + 1:02d}/{epochs}] | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Train Acc: {train_accuracy:.2f}% | "
            f"Val Acc: {val_accuracy:.2f}% | "
            f"LR: {current_lr:.6f}"
        )

    return model, history
````

````python
@torch.inference_mode()
def evaluate(model, loader, device):
    model.eval()
    criterion = nn.CrossEntropyLoss()
    total_loss, total_correct, total_seen = 0.0, 0, 0

    for images, labels in loader:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)
        logits = model(images)
        loss = criterion(logits, labels)
        total_loss += loss.item() * images.size(0)
        total_correct += (logits.argmax(dim=1) == labels).sum().item()
        total_seen += labels.size(0)

    return total_loss / total_seen, total_correct / total_seen


def train_model(model, train_loader, val_loader, optimizer,
                scheduler=None, epochs=20, device=device):
    criterion = nn.CrossEntropyLoss()
    model = model.to(device)
    best_state = copy.deepcopy(model.state_dict())
    best_val_acc = -1.0
    history = {
        "train_loss": [], "train_acc": [],
        "val_loss": [], "val_acc": [], "lr": []
    }
    start = time.time()

    for epoch in range(1, epochs + 1):
        model.train()
        total_loss, total_correct, total_seen = 0.0, 0, 0

        for images, labels in train_loader:
            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)
            optimizer.zero_grad(set_to_none=True)
            logits = model(images)
            loss = criterion(logits, labels)
            loss.backward()
            optimizer.step()

            total_loss += loss.item() * images.size(0)
            total_correct += (logits.argmax(dim=1) == labels).sum().item()
            total_seen += labels.size(0)

        train_loss = total_loss / total_seen
        train_acc = total_correct / total_seen
        val_loss, val_acc = evaluate(model, val_loader, device)
        current_lr = optimizer.param_groups[0]["lr"]

        history["train_loss"].append(train_loss)
        history["train_acc"].append(train_acc)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)
        history["lr"].append(current_lr)

        if scheduler is not None:
            if isinstance(scheduler, optim.lr_scheduler.ReduceLROnPlateau):
                scheduler.step(val_loss)
            else:
                scheduler.step()

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_state = copy.deepcopy(model.state_dict())

        print(
            f"Epoch {epoch:02d}/{epochs} | train loss {train_loss:.4f} | "
            f"train acc {train_acc:.4f} | val loss {val_loss:.4f} | "
            f"val acc {val_acc:.4f} | lr {current_lr:.6g} | "
            f"{time.time() - start:.1f}s"
        )

    model.load_state_dict(best_state)
    print(f"Best validation accuracy: {best_val_acc:.4f}")
    return model, history


def display_history_table(history):
    """Display per-epoch training metrics as a pandas table."""
    table = pd.DataFrame(history)
    table.insert(0, "epoch", np.arange(1, len(table) + 1))
    display(table)
    return table


def plot_training_history(history):
    epochs = range(1, len(history["train_loss"]) + 1)
    plt.figure(figsize=(7, 4))
    plt.plot(epochs, history["train_loss"], label="Train loss")
    plt.plot(epochs, history["val_loss"], label="Validation loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.show()

    plt.figure(figsize=(7, 4))
    plt.plot(epochs, history["train_acc"], label="Train accuracy")
    plt.plot(epochs, history["val_acc"], label="Validation accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.show()
````

### 5.1 Configure and run SimCLR pretraining

Recreate the model for a fresh run so it starts with newly initialized weights.

````python
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    print(f"Available GPU memory: {torch.cuda.mem_get_info()[0] / 1024**3:.2f} GB")
````

````python
set_random_seed(SEED)

EMBED_DIM = 192
NUM_LAYERS = 4
NUM_HEADS = 4
MLP_DIM = 768

# -------------------------
# Create ViT Encoder
# -------------------------
vit_encoder = VisionTransformer(
    img_size=IMG_SIZE,
    patch_size=PATCH_SIZE,
    in_channels=IN_CHANNELS,
    num_classes=num_classes,
    embed_dim=EMBED_DIM,
    num_heads=NUM_HEADS,
    num_layers=NUM_LAYERS,
    mlp_dim=MLP_DIM,
    dropout=DROPOUT
)

# -------------------------
# Create SimCLR (encoder + projection head)
# -------------------------
PROJ_HIDDEN_DIM = 512
PROJ_OUT_DIM = 128
TEMPERATURE = 0.5

simclr_model = SimCLR(
    vit_model=vit_encoder,
    proj_hidden_dim=PROJ_HIDDEN_DIM,
    proj_out_dim=PROJ_OUT_DIM
).to(device)

# -------------------------
# Training Configuration
# -------------------------
LEARNING_RATE = 3e-4
WEIGHT_DECAY = 1e-4
SIMCLR_EPOCHS = 100

optimizer = optim.AdamW(
    simclr_model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=SIMCLR_EPOCHS
)

# -------------------------
# SimCLR Pretraining
# -------------------------
simclr_model, simclr_history = train_simclr(
    simclr_model,
    simclr_train_loader,
    simclr_monitor_loader,
    optimizer,
    scheduler=scheduler,
    epochs=SIMCLR_EPOCHS,
    temperature=TEMPERATURE,
    knn_every=10,
    device=device
)

torch.save(simclr_model.state_dict(), "pretrained_simclr.pth")

plot_simclr_history(simclr_history)
````

### Inspect the learned representation

A good SimCLR encoder should place images of the same class near each other even though it never saw labels. The plot below projects the frozen encoder features of the validation images to 2-D with PCA and colors them by class.

````python
@torch.no_grad()
def visualize_simclr_embeddings(encoder, loader, class_names, device):
    feats, labels = extract_features(encoder, loader, device)

    # PCA to 2-D
    centered = feats - feats.mean(dim=0, keepdim=True)
    _, _, V = torch.pca_lowrank(centered, q=2, center=False)
    coords = (centered @ V[:, :2]).numpy()
    labels = labels.numpy()

    plt.figure(figsize=(8, 6))
    cmap = plt.get_cmap("tab20")
    for class_id, class_name in enumerate(class_names):
        sel = labels == class_id
        plt.scatter(coords[sel, 0], coords[sel, 1], s=12, color=cmap(class_id % 20),
                    label=class_name, alpha=0.8)
    plt.title("PCA of SimCLR encoder features (validation set)")
    plt.xlabel("PC 1")
    plt.ylabel("PC 2")
    plt.grid(alpha=0.25)
    if len(class_names) <= 20:
        plt.legend(fontsize=7, loc="center left", bbox_to_anchor=(1.0, 0.5))
    plt.tight_layout()
    plt.show()

    print(f"Final k-NN validation accuracy: "
          f"{knn_accuracy(encoder, knn_train_loader, knn_val_loader, device):.2f}%")


visualize_simclr_embeddings(simclr_model.encoder, val_loader, class_names, device)
````

### Fine-tune the pretrained encoder for classification

The projection head is discarded and a new linear head is trained on top of the SimCLR encoder. Set `FREEZE_ENCODER = True` for a **linear probe** (a clean measure of representation quality) or `False` to fine-tune the whole network.

````python
EPOCHS       = 90
WARMUP       = 5          # epochs
ENCODER_LR   = 1e-4       # pretrained parts: small
HEAD_LR      = 1e-3       # new head: larger
WEIGHT_DECAY = 0.05
FREEZE_ENCODER = False

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

# Reload the saved SimCLR weights
checkpoint = torch.load(
    "pretrained_simclr.pth",
    map_location=device,
    weights_only=True
)
simclr_model.load_state_dict(checkpoint)

# Build the classifier from the pretrained encoder
classifier = SimCLRClassifier(
    simclr_model=simclr_model,
    num_classes=num_classes
).to(device)

for p in classifier.encoder.parameters():
    p.requires_grad = not FREEZE_ENCODER
for p in classifier.classifier.parameters():
    p.requires_grad = True


def split_decay(params):
    """Weight decay only on >=2-D weights; none on biases, norms, class token, pos-embeddings."""
    decay, no_decay = [], []
    for p in params:
        (no_decay if p.ndim < 2 or p.ndim == 3 else decay).append(p)
    return decay, no_decay


enc_decay, enc_no_decay = split_decay(list(classifier.encoder.parameters()))
head_decay, head_no_decay = split_decay(list(classifier.classifier.parameters()))

param_groups = [
    {"params": head_decay,    "lr": HEAD_LR, "weight_decay": WEIGHT_DECAY},
    {"params": head_no_decay, "lr": HEAD_LR, "weight_decay": 0.0},
]
if not FREEZE_ENCODER:
    param_groups += [
        {"params": enc_decay,    "lr": ENCODER_LR, "weight_decay": WEIGHT_DECAY},
        {"params": enc_no_decay, "lr": ENCODER_LR, "weight_decay": 0.0},
    ]

optimizer = torch.optim.AdamW(param_groups, betas=(0.9, 0.999))


# Per-epoch warmup + cosine. The lambda scales each group's own base LR.
def lr_lambda(epoch):
    if epoch < WARMUP:
        return (epoch + 1) / WARMUP
    progress = (epoch - WARMUP) / max(1, EPOCHS - WARMUP)
    return 0.01 + 0.99 * 0.5 * (1 + math.cos(math.pi * progress))

scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)

classifier, classification_history = train_classifier(
    classifier, train_loader, val_loader, optimizer,
    scheduler=scheduler,
    epochs=EPOCHS, device=device
)
````

````python
# After training: confirm the classifier shares weights with the pretrained SimCLR encoder
print("Encoder shared:", classifier.encoder is simclr_model.encoder)
print("Patch embedding shared:", classifier.encoder.patch_embed is simclr_model.encoder.patch_embed)
print("Position embedding shared:",
      classifier.encoder.position_embedding is simclr_model.encoder.position_embedding)

plot_training_history(classification_history)
````
